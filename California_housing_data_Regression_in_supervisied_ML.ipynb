#Regression

regression-to predict a continuous outplot value (target).
The goal of regression is to find the best modal that accuractly predicts the target value for new unseen inputs.

Different regression methods:-
1. Linear regression method-find the best Linear relationship between the input features and the target values.
2. polynomial regression method-non linear.
3. ridge Regression-Include a regularization term of prevent overfitting.
4. lasso regression- A regularization term to shrink the coefficent of less important features to zero.
5. Decision Tree Regression- rules to make predictions
6. Random forest Regrission
7. Gradian boosting Regression.
